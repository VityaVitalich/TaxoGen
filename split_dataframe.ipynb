```python
import pandas as pd
import random
import os
import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet as wn
from tqdm import tqdm
```

```python
human_def = pd.read_csv('/home/data/v.moskvoretskii/TaxoGen/data/labeling/human_with_flux_def_final.csv')
human_def = human_def.drop(columns=['Unnamed: 0'])
human_def = human_def.sort_values(by=['wordnet_id'])

human_nodef = pd.read_csv('/home/data/v.moskvoretskii/TaxoGen/data/labeling/human_with_flux_nodef_final.csv')
human_nodef = human_nodef.sort_values(by=['wordnet_id'])


gpt_def = pd.read_csv('/home/data/v.moskvoretskii/TaxoGen/data/main_gpt4_def_with_flux.csv')
gpt_nodef = pd.read_csv('/home/data/v.moskvoretskii/TaxoGen/data/main_gpt4_nodef_with_flux.csv')
```

```python
def add_id(df):

    df['id'] = (df['wordnet_id'].astype(str) + df['model_a'] + df['model_b']).apply(hash)
    return df

human_def = add_id(human_def)
human_nodef = add_id(human_nodef)
gpt_def = add_id(gpt_def)
gpt_nodef = add_id(gpt_nodef)
```

```python
print(len(human_def), len(human_nodef))
final = human_def.merge(human_nodef[['result', 'id']], on='id', suffixes=('_human_def', '_human_nodef'))
print(len(final))
```

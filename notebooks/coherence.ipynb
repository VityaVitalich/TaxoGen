```python
merged = ira.merge(vitya, on='model', how='left', suffixes=('_ira', '_vitya'))
merged = ira.merge(vitya, on='model', how='left', suffixes=('_ira', '_vitya'))

merged = merged.merge(kate, on='model', how='left')
```

```python
merged = merged.loc[:, ['model', 'score_ira', 'lower_ira', 'upper_ira', 'score_vitya', 'upper_vitya', 'lower_vitya', 'score', 'lower', 'upper']]
```

```python
merged['score_alina'] = alina['score']
merged['lower_alina'] = alina['lower']
merged['upper_alina'] = alina['upper']
```

```python
spearmanr(merged['lower_vitya'], merged['lower_ira'])
```

```python
types = 'lower'
merged.drop(columns='model').corr()
```

```python
prefix = 'score'
scores_df = merged[[f'{prefix}_ira', f'{prefix}_vitya', prefix, f'{prefix}_alina']]

# Initialize lists to store results
results = []
columns = scores_df.columns

# Compute correlations and p-values between all pairs of scores
for col1, col2 in itertools.combinations(columns, 2):
    if col1 != col2:  # Exclude self-comparison
        corr, p_value = pearsonr(scores_df[col1].dropna(), scores_df[col2].dropna(), method=MonteCarloMethod(n_resamples=10000))
        results.append({'Assessor Pair': f'{col1} vs {col2}', 'Correlation': corr, 'P-Value': round(p_value, 4)})

# Convert results into a DataFrame
correlation_table = pd.DataFrame(results)
```

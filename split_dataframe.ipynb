```python
import pandas as pd
import random
import os
import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet as wn
from tqdm import tqdm
```

```python
df = pd.read_csv('main.tsv', sep='\t')

```

```python
models_list = os.listdir('/home/data/v.moskvoretskii/taxo_demo_images/images/')
```

```python
df['core_synset'][0]
```

```python
df['definition'] = [wn.synset(df['core_synset'][i]).definition() for i in tqdm(range(len(df)))]
```

```python

# target_size = 4 * 1000
# rows_needed = target_size - len(df)
# additional_rows = df.sample(n=rows_needed, replace=True)
# expanded_df = pd.concat([df, additional_rows], ignore_index=True)

#models_list = ['SD3', 'SDXL', 'SDXL_Turbo', 'Kandinsky', 'PixArt_Sigma', 'Playground', 'IF', 'Openjourney', 'Hunyuan-DiT', 'SD_V1.5', 'Retrieved']

def sample_model_pair(models_list):
    return random.sample(models_list, 2)

# Adding the sampled pairs to the dataframe
df[['model_a', 'model_b']] = df.apply(lambda row: pd.Series(sample_model_pair(models_list)), axis=1)
```

```python
import pandas as pd 
import os
```

```python
df = pd.read_csv('/home/data/v.moskvoretskii/TaxoGen/data/main_gpt4_nodef_with_flux.csv')
```

```python
name_of_columns = {
    'stabilityai_sdxl-turbo': 'SDXL-turbo',
    'retrieval': 'Retrieval',
    'runwayml_stable-diffusion-v1-5': 'SD1.5',
    'Tencent-Hunyuan_HunyuanDiT-v1.2-Diffusers': 'HDiT',
    'playgroundai_playground-v2.5-1024px-aesthetic': 'Playground',
    'prompthero_openjourney': 'Openjourney',
    'kandinsky-community_kandinsky-3': 'Kandinsky3',
    'stabilityai_stable-diffusion-xl-base-1.0': 'SDXL',
    'PixArt-alpha_PixArt-Sigma-XL-2-512-MS': 'PixArt',
    'DeepFloyd_IF-I-XL-v1.0': 'DeepFloyd',
    'stabilityai_stable-diffusion-3-medium-diffusers': 'SD3',
    'black-forest-labs_FLUX.1-dev': 'FLUX'
}

subset_rename_mapping = {
    'appendix': 'Easy',
    'appendix_llama': 'P-Easy',
    'leafs_and_no_leafs': 'Hypo',
    'leafs_and_no_leafs_llama': 'P-Hypo',
    'predict_hypernym': 'Hyper',
    'predict_hypernym_llama': 'P-Hyper',
    'simple_triplet_2parent': 'Mix',
    'simple_triplet_2parent_llama': 'P-Mix',
    'mean': 'Mean'
}
```

```python
subsets = list(subset_rename_mapping.keys())
```

```python
# Dictionary to store data for LaTeX table
latex_data = {}

# Iterate over each subset
for subset in subsets:
    new_df_path = f'/home/data/v.moskvoretskii/TaxoGen/data/labeling/human_with_flux_def_final.csv_{subset}_elo.csv'
    
    if os.path.exists(new_df_path):
        new_df = pd.read_csv(new_df_path)
        
        for index, row in new_df.iterrows():
            model = row['model']
            score = row['score']
            upper = row['upper']
            lower = row['lower']
            
            # Calculate deviations
            deviation_plus = upper - score
            deviation_minus = score - lower
            
            # Format the value as required
            value = f"{score:.0f} (+{deviation_plus:.0f}/-{deviation_minus:.0f})"
            
            if model not in latex_data:
                latex_data[model] = {}
            latex_data[model][subset] = value

# Convert the dictionary to a DataFrame
latex_df = pd.DataFrame(latex_data).T

latex_df.rename(columns=subset_rename_mapping, inplace=True)
latex_df.rename(index=name_of_columns, inplace=True)
```

```python
latex_df = latex_df[['Easy', 'Hypo', 'Hyper', 'Mix', 'P-Easy', 'P-Hypo', 'P-Hyper', 'P-Mix']]
```

```python
print(latex_df.to_latex())
```
